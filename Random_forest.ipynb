```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import LabelEncoder
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import mean_squared_error,mean_absolute_error,r2_score
import joblib
```

```python
df=pd.read_csv("student_data.csv")
```

```python
le=LabelEncoder()
df['internet']=le.fit_transform(df['internet'])
df['activities']=le.fit_transform(df['activities'])
df['Mjob']=le.fit_transform(df['Mjob'])
```

```python
df.columns
```

```python
features=['age','traveltime','freetime','health','studytime','activities','internet','absences','G1','G2','failures']
df[features].head()
```

```python
x=df[features]
y=df.G3
```

```python
x_train,x_val,y_train,y_val=train_test_split(x,y,test_size=0.25,random_state=42)
```

```python
model=RandomForestRegressor(n_estimators=10)
model=model.fit(x_train,y_train)
```

```python
model.predict([[13,2,1,2,8,1,1,0,10,15,0]])
```

```python
pred=model.predict(x_val)
acc=r2_score(y_val,pred)
print(acc*100)
mae=mean_absolute_error(y_val,pred)
print(mae)
mse=mean_squared_error(y_val,pred)
print(mse)
```

```python
joblib.dump({
    "model": model,
    "encoder": le
}, "Random_forest.pkl")
```
